## 1. Introduction

Recent events in the Middle East have had a huge international impact. The blockage of the strait of Hormuz, for example, has caused fuel prices to reach peak levels across the globe. Fuel prices are known to be very sensitive to geopolitical changes. Historically, there have been many instances of sudden price increases. 

Rapid fuel price surges can influence travel behaviour. For example, increasing conventional fuel prices could lead to an increase in EV ownership, or to transport modal shifts. The main objective of this project was to analyse how changes in fuel prices compare to trips with alternative transport options across European countries. To narrow down the scope, the research investigated this for the number of passengers using railway transport, as this mode could be considered the main competitor for car transport in short and medium distances. Furthermore, passenger railway services are widely available in the European Union, allowing a comparison between Member States.

## 2. Research Question

**2.1 Main Question:**

**How has railway passenger demand moved alongside fuel price changes in European countries between 2006 and 2025 and how do these patterns differ between countries?**

**2.2 Sub-questions:** 

1. How did rail passenger demand and real fuel prices differ between European countries over 2006-2025?
2. How did year-on-year changes (quarterly) in real fuel prices and rail passenger demand develop during the main fuel price episodes (2008, 2014-2016, and 2022)?
3. How strongly are rail passenger demand and real fuel prices correlated in each country?
4. How strongly does rail demand respond to fuel price changes in the same quarter?
5. Which countries show the strongest association and what characteristics do these countries share?

**3.2 Geographical and Temporal Scope** 

The geographical scope of the study covers EU Member States for which comparable railway passenger data are available. Five countries have been excluded from the analysis due to lack of railway passenger services (Cyprus and Malta) and due to insufficient data for the selected dependent variable (Austria, Belgium and Hungary).

<img src="images/geographical_scope.png" alt="Geographical Scope" style="display: block; margin: 0 auto" width="50%">

The temporal scope of the study spans a twenty-year period, from 2006 until 2025. This period includes several major geopolitical events influencing fuel prices, such as the 2008 financial crisis, several military conflicts and COVID-19.

**3.3 Data preparation and methods**

In order to harmonise data for the analysis, datasets were cleaned and combined in the script. All series were restricted to 2006-2025 and to the 22 selected countries. From the Oil Bulletin, header rows and rows with an invalid date were removed, and only Euro95 prices including taxes were kept. Any non-numeric values were set to missing. Weekly fuel prices and monthly price indices were then averaged per quarter. For the population, due to the annual reporting in the source, the value was repeated for each quarter in a year. Rail passengers (in thousands) were then expressed per capita by dividing by the population to account for differences in country size and population changes. Fuel prices (per 1000 liters) were converted to euros per liter, consequently dividing nominal prices by the price index (2025 = 100), resulting in real fuel prices.

Missing values were not imported. Correlation and regressions used exclusively observations with both variables available, with no weight attributed to countries with missing data in the weighted averages. The analysis consisted of (1) population-weighted year-over-year changes and event comparisons, (2) correlations per country and (3) log-log regressions with country fixed effects and HAC standard errors. The main specification excludes 2020 Q1 - 2023 Q1 since COVID-19 simultaneously affected rail demand and fuel prices, as changes in passenger trips could otherwise be associated with price fluctuations.

**5.2 Regression analysis**

The event analysis and correlations show that rail demand and fuel prices moved together in some periods. This regression tests whether this relationship holds across all quarters and countries. The year-on-year log change in rail passengers per capita is regressed on the year-on-year log change in the real Euro95 price, so the coefficient can be read as an elasticity. 

As mentioned in Chapter 3, the main analysis excludes 2020Q1 to 2023Q1. The pandemic affected both variables at the same time, so including it would create a relationship unrelated to how travellers respond to fuel prices. Results including COVID-19 are also shown for comparison.

## 6. Conclusion

In a global context, shaped by economic and geopolitical factors, fuel prices experience fluctuations, which could affect travel patterns. This project used data from 22 European countries to assess how rail passenger demand moved alongside fuel price changes between 2006 and 2015, and how do these patterns differ between countries. Overall, the results suggest that no consistent, strong relationship between the two variables could be established.

**6.1 Changes during the fuel price episodes**

At an aggregate level, rail demand did not respond consistently to fuel price windows analysed. During the 2008 financial crisis (average -1.2%), railway passenger trips per capita declined despite lower fuel prices, shaped by the recession which affected incomes and travel patterns overall. During the Crimea crisis, rail demand continued to grow (+1.0%), although fuel was 6.1% cheaper. The only instance of simultaneous increases was observed in 2022, reflecting the post-pandemic recovery alongside the effects of the war in Ukraine.

**6.2 Differences between countries in levels**

Cross-country similarity in timing was found to be strong for both variables, as the EU Member States of the EU likely experience similar consequences of shared global and economic events. The main differences are in level; for example, Bulgaria and the Netherlands have a track record of the lowest and highest real fuel prices, respectively, with values ranging by 0.60 and 0.80 eurocents on average.

**6.3 Correlations per country**

Rail demand and real fuel prices are weakly to moderately correlated, ranging in sign and across countries, from -0.391 (Slovakia) to +0.396 (Bulgaria), suggesting a lack of uniform relationship and a possible influence of local factors such as rail availability, car dependency, which also differ across the Member States.

**6.4 Same-quarter response**

The pooled regression analysis yielded an elasticity of -0.032, which is statistically insignificant (p = 0.35), excluding the COVID-19 quarters. This means that a 10% increase in real fuel prices corresponds to a shift in rail demand of approximately -1.0% to +0.35%. If the pandemic quarters are considered, the elasticity rises significantly to 0.80. However, it could not be concluded that travellers prefer a different mode, but a result of the COVID-19 restrictive measures and lower demand for transport services overall. The link between fuel prices and rail demand changes over time. Large and sudden price shocks may push some travellers towards rail, but in most quarters the effect is weak.

**6.5 Strongest associations among countries**

Bulgaria was found to have the strongest positive association (r = 0.396; elasticity +0.141), while Slovakia shows the most negative correlation (-0.391) and Croatia the strongest negative elasticity (-0.636). Contrastingly, Croatia, Greece, Ireland and Portugal, all of which were severely economically affected after 2008, have the most negative elasticities.

Overall, changes in rail demand per capita predominantly followed economic conditions, restrictions and recovery in mobility, rather than fuel prices explicitly. This is likely because fuel is only part of the cost of driving or because railway alternatives are not attractive for many trips. Furthermore, the global oil market dictates fluctuations in fuel prices. Finally, there are no substantial relationships between the variables across the countries investigated. The resulting associations are not causal effects.


## 7. Discussion

**7.1 Interpretation of the findings**

The weak relationship between real fuel prices and rail passenger demand per capita is a reasonable outcome. Besides increasing prices in petrol stations across Europe, environmental regulations and congestion zoning are also likely causes for changing travel patterns. Nonetheless, policy alone might not necessarily be effective in convincing travellers to switch to alternative modes, as they might be inconvenient, particularly in rural areas with poor public transport connectivity.

Changes in quarterly fuel prices are often too small to change travel habits. Instead, economic growth is usually associated with higher prices, while recessions, such as following the 2008 financial crisis, shrink both travel demand and fuel prices. For instance, the four countries with the most negative elasticities (Croatia, Greece, Ireland and Portugal) experienced lower travel demand amid economic crisis, which likely affected railway services as well. On the other hand, Bulgaria's highest positive association could reflect the lower car ownership rate than the EU average and higher price sensitivity.


**7.2 Limitations and future research**

This analysis has several limitations to be considered. Firstly, rail passenger data have gaps in some countries and years, and the peak of +18% in 2012 Q4 to -0.9% afterwards is likely a result of changing coverage than a real surge, considering the same effect is not observed for other countries. Additionally, the lack of data for five of the EU Member States leaves the conclusions incomplete, while the limitation of available datasets for all variables prior to 2006 make single-country estimates imprecise. Therefore, future research could address these gaps through the addition of economic and policy variables, such as income and unemployment data, GDP, subsidy for public transport, etc. This could provide additional insights on the travel patterns of people around the EU in the long-term.
