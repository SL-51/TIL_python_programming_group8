## 1. Introduction

Recent events in the middle east have had a huge international impact. The blockage of the strait of Hormuz by Iran for example has caused fuel prices to reach peak levels across the globe. Oil prices are known to be very sensitive to geopolitical changes. It is however not the first time that oil prices spike like they do now. Historically there have been many events when oil prices have suddenly peaked. 

Rapid fuel price inflations can lead to a difference in travel behaviour. For example, a spike in electric cars, or people shifting towards other modes of transport. This project aims to analyse how spikes in fuel prices compare to travel behaviour pattern changes for the European Union. To narrow down the scope the research investigates this for number of passengers using railway transport. Railway transport has been chosen, because it is a widely used mode of transport in the European union and additionally, it is considered the main competitor for car transport.

## 2. Research Question

**2.1 Main Question:** How has railway passenger demand moved alongside fuel price changes in European countries between 2006 and 2025 and how do these patterns differ between countries?

**2.2 Sub-questions:** 

1. How did rail passenger demand develop during the main fuel price episodes (2008, 2014-2016, and 2022)?
2. How strongly are year-on-year changes (quarterly) in real fuel prices and rail passenger demand correlated in each country?
3. Does rail demand respond to fuel price changes immediately or with a delay of one or more quarters?
4. Which countries show the strongest association and what characteristics do these countries share?

**5.2 Regression analysis**

The event analysis and correlations show that rail demand and fuel prices moved together in some periods. This regression tests whether this relationship holds across all quarters and countries. The year-on-year log change in rail passengers per capita is regressed on the year-on-year log change in the real Euro95 price, so the coefficient can be read as an elasticity. 
The main analysis excludes 2020Q1 to 2023Q1. The pandemic affected both variables at the same time, so including it would create a relationship unrelated to how travellers respond to fuel prices. Results including COVID-19 are also shown for comparison.

The pooled elasticity is −0.032 and not statistically significant (p = 0.35) without COVID-19. Changes in fuel prices are not linked to noticeable changes in rail demand across Europe.
The elasticity rises to 0.80 and becomes highly significant with COVID-19 included but this does not mean travellers switched modes. The pandemic quarters are only a small part of the data, yet they change the result completely, and all 22 countries show similar estimates despite very different rail systems. 

| Specification | Elasticity | 95% CI | p-value | Observations |
|:---|---:|:---:|---:|---:|
| Excluding COVID-19 | -0.032 | [-0.100, 0.035] | 0.348 | 1,329 |
| Including COVID-19 | 0.804 | [0.651, 0.957] | <0.001 | 1,610 |

Most elasticities are close to zero at the country level and only Croatia and Bulgaria are statistically significant. 14 of the 22 estimates are negative and the four most negative belong to Croatia, Greece, Ireland and Portugal. All four went through severe economic crises after 2008 which suggests that economic conditions rather than fuel prices shaped rail demand there.
Bulgaria is the only country where rail demand clearly rose with fuel prices. The effect is small as a 10% fuel price increase goes with about 1.4% more rail passengers.
Croatia is the only country where rail demand clearly fell when fuel prices rose. There is no reason why people would take the train less because driving became more expensive so this result most likely reflects other developments in the same period. Croatia went through a long recession after 2008 in which overall travel declined and changes in rail services may also have played a role. The result is therefore not evidence against the expected relationship but an example of how wider economic conditions can outweigh the effect of fuel prices.

## 7. Conclusion

Fuel is only part of the cost of driving. Car owners have already paid for the car, insurance and taxes so a higher fuel price raises the total cost of a trip by much less and driving often stays the easier option. The train is not a realistic alternative for many trips, especially in rural areas or with several stops, and drivers tend to adjust in simpler ways first, such as driving less or carpooling. Most quarterly fuel price changes are also too small to change travel habits. The economy plays a role as well. Fuel prices are usually high when the economy is doing well and people travel more and drop during recessions which can hide or even reverse a shift towards rail.

The link between fuel prices and rail demand changes over time. Large and sudden price shocks may push some travellers towards rail, but in most quarters the effect is weak. Fuel prices seem to be just one of the things that shape rail demand, how practical the train is and how the economy matters more.
