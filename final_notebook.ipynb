## 1. Introduction

Recent events in the middle east have had a huge international impact. The blockage of the strait of Hormuz by Iran for example has caused fuel prices to reach peak levels across the globe. Oil prices are known to be very sensitive to geopolitical changes. It is however not the first time that oil prices spike like they do now. Historically there have been many events when oil prices have suddenly peaked. 

Rapid fuel price inflations can lead to a difference in travel behaviour. For example, a spike in electric cars, or people shifting towards other modes of transport. This project aims to analyse how spikes in fuel prices compare to travel behaviour pattern changes for the European Union. To narrow down the scope the research investigates this for number of passengers using railway transport. Railway transport has been chosen, because it is a widely used mode of transport in the European union and additionally, it is considered the main competitor for car transport.

## 2. Research Question

**2.1 Main Question:** How has railway passenger demand moved alongside fuel price changes in European countries between 2006 and 2025 and how do these patterns differ between countries?

**2.2 Sub-questions:** 

1. How did rail passenger demand develop during the main fuel price episodes (2008, 2014-2016, and 2022)?
2. How strongly are year-on-year changes (quarterly) in real fuel prices and rail passenger demand correlated in each country?
3. Does rail demand respond to fuel price changes immediately or with a delay of one or more quarters?
4. Which countries show the strongest association and what characteristics do these countries share?

**3.1 Datasets** 

All data is publicly available and standarized across EU countries. The analysis uses quarterly data from 2006 to 2025. Rail passenger demand is expressed per inhabitant to account for differences in country size and population changes. Factors such as rail service quality, car dependency, income levels and the availability of alternative transport modes may also contribute to differences in rail demand. Nevertheless, their influence cannot be established through correlation analysis alone. Furthermore, the coefficients describe associations between the observed values of the variables and do not imply causation. Since the analysis considers the variables in levels, the results may also differ from those obtained when examining changes in rail demand and fuel prices, as explored in the subsequent regression analysis.
